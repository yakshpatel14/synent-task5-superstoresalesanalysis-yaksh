The objective of this project is to analyze business sales performance using the Superstore Sales dataset.

7. Final Conclusion

## 1. import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. load dataset

```python
df = pd.read_csv("Sample - Superstore.csv", encoding='latin1')
df.head()
```

## 3. understand dataset

```python
df.info()
df.describe()
```

## 4. monthly revenue


```python

df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Month'] = df['Order Date'].dt.to_period('M')
monthly_sales = df.groupby('Month')['Sales'].sum()
monthly_sales.plot(figsize=(12,5))
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

## 5. top selling product

```python
top_products = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False).head(10)
top_products.plot(kind='bar', figsize=(10,5))
plt.title("Top Selling Products")
plt.xlabel("Sub-Category")
plt.ylabel("Sales")
plt.show()
```

## 6. profit analysis

```python
profit_analysis = df.groupby('Category')['Profit'].sum()
profit_analysis.plot(kind='pie', autopct='%1.1f%%', figsize=(7,7))
plt.title("Profit Distribution by Category")
plt.ylabel("")
plt.show()
```

## 7. Final Conclusion

The Superstore Sales dataset was successfully analyzed to understand business performance.

### Results
- Monthly revenue trends were identified.
- Top-selling product categories were analyzed.
- Profit contribution by category was visualized.
